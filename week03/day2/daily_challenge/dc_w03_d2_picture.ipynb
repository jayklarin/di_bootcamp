Flip an image horizontally and then vertically using mirror and flip functions from ImageOps


```python
from PIL import ImageOps

def flip_image(image):
    # Flip horizontally (mirror left–right)
    flipped_h = ImageOps.mirror(image)

    # Flip vertically (top–bottom)
    flipped_v = ImageOps.flip(image)

    return flipped_h, flipped_v
```

```python
original = Image.open("flowers/flowers/19_010.png")
fliph, flipv = flip_image(original)

plt.subplot(1,2,1)
plt.title("Horizontal Flip")
plt.imshow(fliph)
plt.axis("off")

plt.subplot(1,2,2)
plt.title("Vertical Flip")
plt.imshow(flipv)
plt.axis("off")
plt.show()
```

Zoom in on an image (scale by 1.2x) using .resize(...)

```python
def zoom_image(image, scale=1.2):
    w, h = image.size
    new_size = (int(w * scale), int(h * scale))
    return image.resize(new_size)
```

```python
zoomed = zoom_image(original, scale=1.2)

plt.imshow(zoomed)
plt.title("Zoomed 1.2×")
plt.axis("off")
plt.show()
```
